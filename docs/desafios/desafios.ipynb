### Desafio 3: classificador de desempenho

Receba o valor vendido e classifique:

menos de R$ 5.000     → BAIXO
R$ 5.000–9.999        → REGULAR
R$ 10.000–19.999      → BOM
R$ 20.000 ou mais     → EXCELENTE

Use if, elif e else.

Bônus: adicione uma meta e apresente também a porcentagem atingida.

```python
valor_venda = float(input('Informe o valor da venda em R$: \n-> '))
meta = float(input('Informe uma meta a ser atingida em R$: \n-> '))
porcentagem = (valor_venda/meta) * 100

if valor_venda < 5000 :
    print('Venda Baixa')
elif valor_venda >= 5000 and valor_venda <= 9999 :
    print('Venda Regular')
elif valor_venda > 9999 and valor_venda < 19999 :
    print('Venda Boa')
elif valor_venda >= 20000 :
    print('Venda Exelente')
else :
    print('Valor inválido')

if porcentagem == 100:
    print('Voce bateu a meta, exatamente 100%')
elif porcentagem > 100:
    print(f'Venda: {valor_venda} \nMeta: {meta}\n Voce passou da meta: Passou {porcentagem:.1f - 100}%')
else :
    print(f'Venda: {valor_venda} \nMeta: {meta}\nMeta não atingida: falta {100 - porcentagem:.1f}% para atingir a meta')
```

```python
vendas = [7200, 8500, 4900, 12000, 15700, 6300]
acumulador = 0
soma = 0
for venda in vendas:
    if venda > 10000:
        print(venda)
        acumulador = acumulador + 1
    elif venda <= 10000:
        soma = soma + venda

            
print(soma)
print(acumulador)
```

### Desafio 5: calculadora comercial

Transforme os cálculos anteriores em funções.

Você deve criar, por exemplo:

calcular_comissao()
calcular_total()
calcular_media()
verificar_meta()
classificar_desempenho()

Não estou te dando a implementação, somente a responsabilidade de cada função.

O programa principal deve chamá-las.



```python
# Calcular Medias | Verificar Meta | Classificação de Desempenho
# Meta: fazer o usuario colocar a quantidade de venda que ele quiser. e continuar o programa

salario_base = input('informe o salario base: ')
if salario_base == '':
    salario_base = 1800
salario_base = int(salario_base)
meta = int(input('defina uma meta mensal: '))
COMISSAO = 4/100
lista_vendas = []

def salario_tratado(lista_vendas):
    salario_total = (sum(lista_vendas) * COMISSAO) + salario_base
    return salario_total
def calcular_media(lista_vendas):
    qtd_vendas = len(lista_vendas)
    media = sum(lista_vendas) / qtd_vendas
    return media
def verificacao_por_meta(lista_vendas,meta):
    salario_por = salario_tratado(lista_vendas)
    porcentagem_meta = (salario_por / meta) * 100
    return porcentagem_meta
# Se a soma de vendas >= que a meta : Bateu a meta! -> USAR 
def verificacao_meta(lista_vendas,meta):
    por_meta = verificacao_por_meta(lista_vendas=lista_vendas,meta=meta)
    salario_final = salario_tratado(lista_vendas)
    if meta > salario_final:
        print(f'Seu salario final não alcançou a sua meta mensal estabelecida, falta {por_meta}% para alcançar a meta')
    elif meta <= salario_final:
        print(f'Seu salario final Alcançou a meta mensal estabelecida, {por_meta}%')
    else:
        print('Valor Invalido')
# Classificação de Desempenho -> USAR
def classificacao_media_sal(lista_vendas):
    media_venda = calcular_media(lista_vendas)
    if media_venda < 5000 :
        print('Venda Media Baixa')
    elif media_venda >= 5000 and media_venda <= 9999 :
        print('Venda Media Regular')
    elif media_venda > 9999 and media_venda < 19999 :
        print('Venda Media Boa')
    elif media_venda >= 20000 :
        print('Venda Media Exelente')
    else :
        print('Valor de media inválida')    

# enquanto o usuario digitar uma venda , adicionar a iteração na lista vazia
valor = 0
numero_venda = 1
vendas = input('Digite todas as vendas realizada no mês, existe vendas? sim = continue , nao = O programa para \n-> ')
while vendas == 'sim':
    valor = int(input(f'Qual {numero_venda}º valor?: '))
    lista_vendas.append(valor)
    numero_venda = numero_venda + 1
    vendas = input('deseja continuar? sim ou nao?\n-> ')
    while vendas != 'sim' and vendas != 'nao': 
        vendas = input('deseja continuar? sim ou nao?\n-> ')         
    
salario = salario_tratado(lista_vendas=lista_vendas)
verificacao_meta(lista_vendas=lista_vendas,meta=meta)
classificacao_media_sal(lista_vendas=lista_vendas)
print(f'Você tem {len(lista_vendas)} vendas, são elas: {lista_vendas}, seu salario total ficará : {salario}')

```
