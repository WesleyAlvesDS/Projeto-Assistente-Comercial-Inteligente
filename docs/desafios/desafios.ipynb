### Desafio 3: classificador de desempenho

Receba o valor vendido e classifique:

menos de R$ 5.000     → BAIXO
R$ 5.000–9.999        → REGULAR
R$ 10.000–19.999      → BOM
R$ 20.000 ou mais     → EXCELENTE

Use if, elif e else.

Bônus: adicione uma meta e apresente também a porcentagem atingida.

```python
valor_venda = float(input('Informe o valor da venda em R$: \n-> '))
meta = float(input('Informe uma meta a ser atingida em R$: \n-> '))
porcentagem = (valor_venda/meta) * 100

if valor_venda < 5000 :
    print('Venda Baixa')
elif valor_venda >= 5000 and valor_venda <= 9999 :
    print('Venda Regular')
elif valor_venda > 9999 and valor_venda < 19999 :
    print('Venda Boa')
elif valor_venda >= 20000 :
    print('Venda Exelente')
else :
    print('Valor inválido')

if porcentagem == 100:
    print('Voce bateu a meta, exatamente 100%')
elif porcentagem > 100:
    print(f'Venda: {valor_venda} \nMeta: {meta}\n Voce passou da meta: Passou {porcentagem:.1f - 100}%')
else :
    print(f'Venda: {valor_venda} \nMeta: {meta}\nMeta não atingida: falta {100 - porcentagem:.1f}% para atingir a meta')
```

Transforme os cálculos anteriores em funções.

Não estou te dando a implementação, somente a responsabilidade de cada função.

O programa principal deve chamá-las.



```python
# Calcular Medias | Verificar Meta | Classificação de Desempenho
# Meta: fazer o usuario colocar a quantidade de venda que ele quiser. e continuar o programa

COMISSAO = 4/100 # valor ajustavel
meta_padrao = 2000 # valor ajustavel, poderá ser um valor da meta mensal estabelecido pela empresa
salario_padrao = 1800 # valor ajustavel
salario_base = input(f'Informe o salario base: padrão = {salario_padrao}R$')
if salario_base == '':
    salario_base = salario_padrao
salario_base = int(salario_base)
meta = input(f'Defina uma meta de vendas: padrão {meta_padrao}R$')
if not meta:
    meta = meta_padrao
meta = int(meta)
lista_vendas = []

def salario_tratado(lista_vendas):
    salario_total = (sum(lista_vendas) * COMISSAO) + salario_base
    return salario_total
def vendas_somadas(lista_vendas):
    soma = sum(lista_vendas)
    return soma 
def vendas_media(lista_vendas):
    qtd_vendas = len(lista_vendas)
    media = vendas_somadas(lista_vendas=lista_vendas) / qtd_vendas
    return media
def verificacao_vendas_meta(lista_vendas,meta):
    soma_venda = vendas_somadas(lista_vendas)
    porcentagem_meta = (soma_venda / meta) * 100
    return porcentagem_meta

# Se a soma de vendas >= que a meta : Bateu a meta! -> USAR 
def classificacao_vendas_meta(lista_vendas,meta):
    por_meta = verificacao_vendas_meta(lista_vendas=lista_vendas,meta=meta)
    falta_meta = (100 - por_meta )
    vendas_final = vendas_somadas(lista_vendas)
    excedente = (por_meta - 100)
    if meta > vendas_final:
        return(f'Suas vendas não alcançaram a meta estabelecida, voce alcançou {por_meta}%, ainda falta {falta_meta}% para alcançar a meta')
    elif meta == vendas_final:
        return(f'Suas vendas bateram a meta estabelecida, {por_meta}%')
    elif meta < vendas_final:
        if por_meta > 100:
            return(f'Suas vendas excederam a meta estabelecida, o valor passou a meta em {excedente}%')    
    else:
        return('Vandas não prenchidas')
    
# Classificação de Desempenho -> USAR
def classificacao_media_sal(lista_vendas):
    media_venda = vendas_media(lista_vendas)
    if media_venda < 5000 :
        return('Venda Media Baixa')
    elif media_venda >= 5000 and media_venda <= 9999 :
        return('Venda Media Regular')
    elif media_venda > 9999 and media_venda < 19999 :
        return('Venda Media Boa')
    elif media_venda >= 20000 :
        return('Venda Media Exelente')
    else :
        return('Valor de media inválida')    

# enquanto o usuario digitar uma venda , adicionar a iteração na lista vazia
numero_venda = 1
verifica_vendas = input('Existe vendas? sim = continua , nao = não continua , em seguida digite todas as vendas realizada no mês \n-> ').lower().strip()
while verifica_vendas != 'sim' and verifica_vendas != 'nao':
    verifica_vendas = input('deseja continuar? sim ou nao?\n-> ').lower().strip()
while verifica_vendas == 'sim':
    valor = int(input(f'Qual {numero_venda}º valor?: '))
    lista_vendas.append(valor)
    numero_venda = numero_venda + 1
    verifica_vendas = input('deseja continuar? sim ou nao?\n-> ')
    while verifica_vendas != 'sim' and verifica_vendas != 'nao':
        verifica_vendas = input('deseja continuar? sim ou nao?\n-> ')
if lista_vendas:
    salario = salario_tratado(lista_vendas=lista_vendas)
    desempenho_meta = classificacao_vendas_meta(lista_vendas=lista_vendas,meta=meta)
    desempenho_meta_sal = classificacao_media_sal(lista_vendas=lista_vendas)
    vendas_s = vendas_somadas(lista_vendas=lista_vendas)
    vendas_m = vendas_media(lista_vendas=lista_vendas)

    print(f'Você tem {len(lista_vendas)} vendas, são elas: {lista_vendas}')
    print(f'Seu salario total ficará : {salario}')
    print(f'A soma das suas vendas é:{vendas_s}')
    print(f'A media das suas vendas é:{vendas_m}')
    print(f'{desempenho_meta}')
    print(f'E o desempenho da media de vendas é:{desempenho_meta_sal}')
```
